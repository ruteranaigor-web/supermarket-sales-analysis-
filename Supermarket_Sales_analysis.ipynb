```python
sales_by_product = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
sns.barplot(x=sales_by_product.values, y=sales_by_product.index)

plt.title('Total Sales by Product Line')
plt.xlabel('Total Sales')
plt.ylabel('Product Category')
plt.tight_layout()
plt.show()
```

```python
branch_sales = df.groupby('Branch')['Total'].sum().sort_values(ascending=False)

plt.figure(figsize=(7, 5))
sns.barplot(x=branch_sales.index, y=branch_sales.values)

plt.title('Total Sales by Branch')
plt.xlabel('Branch')
plt.ylabel('Total Sales')
plt.show()
```

```python
payment_sales = df.groupby('Payment')['Total'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=payment_sales.index, y=payment_sales.values)

plt.title('Total Sales by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Sales')
plt.show()
```

```python
monthly_sales = df.groupby(df['Date'].dt.month)['Total'].sum()

plt.figure(figsize=(9, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks([1, 2, 3], ['January', 'February', 'March'])
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(df['Total'], bins=20, edgecolor='black')

plt.title('Distribution of Sales Transactions')
plt.xlabel('Sales Amount')
plt.ylabel('Number of Transactions')

plt.show()
```

```python

df['Day'] = df['Date'].dt.day_name()


days_order = ['Monday', 'Tuesday', 'Wednesday',
              'Thursday', 'Friday', 'Saturday', 'Sunday']

sales_heatmap = df.pivot_table(
    values='Total',
    index='Day',
    columns='Hour',
    aggfunc='sum'
)

sales_heatmap = sales_heatmap.reindex(days_order)
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(sales_heatmap, cmap='YlOrRd')

plt.title('Supermarket Sales by Day and Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Day of the Week')

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(df['Quantity'], df['Total'], alpha=0.6)

plt.title('Relationship Between Quantity Purchased and Total Sales')
plt.xlabel('Quantity of Products')
plt.ylabel('Total Sales')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(y=df['Total'])

plt.title('Distribution of Supermarket Sales')
plt.ylabel('Total Sales')

plt.show()
```

```python
Q1 = df['Total'].quantile(0.25)
Q3 = df['Total'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
high_outliers = df[df['Total'] > upper_limit]

print("Number of high outliers:", len(high_outliers))

high_outliers[['Invoice ID', 'Branch', 'Product line', 'Quantity', 'Total']]
```

```python

print("Average high outlier sales:", high_outliers['Total'].mean())

print("Highest transaction:", high_outliers['Total'].max())

print(high_outliers['Product line'].value_counts())
```

```python
customer_sales = df.groupby('Customer type')['Total'].sum()

print(customer_sales)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(x=customer_sales.index, y=customer_sales.values)

plt.title('Total Revenue by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Total Revenue')

plt.show()
```

```python
average_spending = df.groupby('Customer type')['Total'].mean()

print(average_spending)
```

```python
customer_frequency = df['Customer type'].value_counts()

print(customer_frequency)
```

```python
gender_sales = df.groupby('Gender')['Total'].sum()

print(gender_sales)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(x=gender_sales.index, y=gender_sales.values)

plt.title('Total Revenue by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Revenue')

plt.show()
```

```python
gender_products = pd.crosstab(
    df['Product line'],
    df['Gender']
)

print(gender_products)
```

```python
gender_products.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Product Preferences by Gender')
plt.xlabel('Product Line')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=45)
plt.legend(title='Gender')
plt.tight_layout()
plt.show()
```
